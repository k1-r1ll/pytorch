```python
import torch
```

```python
def act(x):
    return 0 if x < 0.5 else 1
```

```python
def go(house, music, attr):
    X = torch.tensor([house, music, attr], dtype=torch.float32)
    Wh = torch.tensor([[0.3, 0.3, 0],[0.4, -0.5, 1]], dtype=torch.float32)
    Wout = torch.tensor([-1, 1], dtype=torch.float32)

    Zh = torch.mv(Wh, X)
    print(f"Значение сумм на нейронах скрытого слоя: {Zh}")

    Uh = torch.tensor([act(elem) for elem in Zh], dtype=torch.float32)
    print(f"Значение на выходах нейронов скрытого слоя: {Uh}")

    Zout = torch.dot(Wout, Uh)
    Y = act(Zout)
    print(f"Выходное значение НС: {Y}")

    return Y
```

```python
house, rock, attr = 1, 0, 1
```

```python
res = go(house, rock, attr)
print("Yes" if res else "No")
```
