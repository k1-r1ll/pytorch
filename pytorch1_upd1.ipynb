```python
class RandomDataset(torch.utils.data.Dataset):

    def __init__(self, x, y):
        self.x = x
        self.x = y

    def __len__(self):
        return len(self.x)

    def __getitem__(self, idx):
        return {"sample": torch.tensor(x[idx, :], dtype=torch.float), "target": y[idx]}       
```

```python
from torch import nn
```

```python
class SixSevenNet(nn.Module):

    # инициализируем слои
    def __init__(self, in_features, hidden_features, out_features):
        # инициализируем функцию-конструктор из модуля
        super().__init__()
        # объявляем модель и что мы можем в ней подобовлять
        self.model = nn.Sequential(

            # линейный слой (простое уравнение регрессии)
            # тут сколько всего на входе и сколько нейронов хотим получить на выходе
            nn.Linear(in_features, hidden_features),
            # Функция активации ReLU (Rectified Linear Unit), превращающая все отрицательные числа в ноль,
            # а положительные оставляющую без изменений. Формула функции выглядит так: f(x) = max(0, x)
            nn.ReLU(),
            # убираем 20% слоев, чтобы избежать переобучения; регуляризация
            nn.Dropout(0.2),
            # еще один линейній слой, но на выходе hidden_features с вдвое уменьшенной размерностью
            nn.Linear(hidden_features, hidden_features // 2),
            # то же, что и выше, но с небольшим углом, чтобы зануление шло неполностью
            nn.LeakyReLU(0.1),
            # последний слой
            nn.Linear(hidden_features // 2, out_features)
            
        )

    # прогоняем по слоям
    def forward(self, x):
        # распаковываем элемент в матрицу, начиная со второго элемента (первый -- не координата)
        h = torch.flatten(x, start_dim = 1)
        return self.model(h)
```

```python
model = SixSevenNet(784, 128, 10)
```
